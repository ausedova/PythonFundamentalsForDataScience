```python
#Subsetting

unadjusted_business_profits = us_statistics["Unadjusted Business Profits"]

unadjusted_business_profits.head()
```

```python
#Performing a categorization (boolean outcome)

us_statistics["Unadjusted Business Profits"] < 0
```

```python
#Performing a filtered selection (any given outcome)

us_statistics[us_statistics["Unadjusted Business Profits"] < 0]
```

```python
#Performing a filtered multiple selection

us_statistics[(us_statistics["Year"] >= 1929) & (us_statistics["Year"] <= 1939)]
```

```python
us_statistics.loc[(us_statistics["Year"] >= 1929) & (us_statistics["Year"] <= 1939)]

#This can be done equivalently using the loc operator
```

```python
#Performing a filtered multiple selection using the iloc operator

us_statistics.iloc[0:11]

#REMINDER:

#We use loc for label-based filtering (using row/column names)

#We use iloc for position-based filtering (using table positions)
```

```python
#Creating a new variable derived from the information contained in the previously existing variables

us_statistics["Net Economic Surplus"] = (
    us_statistics["Adjusted Business Profits"] - us_statistics["Adjusted Depreciation Estimates"]
)

us_statistics.info()
us_statistics.head()
```

```python
#Renaming a column

us_statistics = us_statistics.rename(
    columns={
        "Year": "Year of Data Analysis"
    }
)

us_statistics.info()
```

```python
#Calculating the mean

us_statistics["Net Economic Surplus"].mean()
```

```python
#Calculating the median

us_statistics["Net Economic Surplus"].median()
```

```python
#Calculating aggregate statistics

us_statistics.describe()
```

```python
#Sorting the data in ascending order by Unadjusted Business Profits

us_statistics.sort_values(by="Unadjusted Business Profits").head()
```

```python
#Sorting the data in descending order by Unadjusted Business Profits

us_statistics.sort_values(by="Unadjusted Business Profits", ascending=False).head()
```

```python
#Wide to long format (using .melt)

us_statistics_long = us_statistics.melt(
    value_vars=["Unadjusted Business Profits", "Unadjusted Depreciation Estimates"],
    var_name="Component",
    value_name="Value",
)

us_statistics_long.info()
us_statistics_long.head()

#We use .pivot for long to wide transformation
```
