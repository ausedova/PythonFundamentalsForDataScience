```python
#pip install polars
#pip install fastexcel

import polars as pl

us_statistics = pl.read_excel("C:/Users/artem/Desktop/Data Science/US Statistics.xlsx")

#Note that we are using pd.read_excel() instead of pd.read_csv() here

print(us_statistics)
```

```python
#Performing a categorization (boolean outcome)

us_statistics.get_column("Unadjusted Business Profits") < 0
```

```python
#Performing a filtered selection (any given outcome)

us_statistics.filter(pl.col("Unadjusted Business Profits") < 0)
```

```python
#Performing a filtered multiple selection

us_statistics.filter(
(pl.col("Year") >= 1929) & (pl.col("Year") <= 1939))

#Filter is the Polars equivalent to .loc / .iloc in Pandas
```

```python
#Performing a multiple selection (without filtering)

us_statistics.select(pl.col(["Year", "Unadjusted Business Profits", "Unadjusted Depreciation Estimates"]))
```

```python
#Indexing data

us_statistics.slice(0, 11)
```

```python
#Creating a new variable derived from the information contained in the previously existing variables

#Please note that .with_column() has been deprecated; use .with_columns() instead

us_statistics = us_statistics.with_columns(
    (
        pl.col("Adjusted Business Profits")
        - pl.col("Adjusted Depreciation Estimates")
    ).alias("Net Economic Surplus")
)

print(us_statistics)
```

```python
#Renaming a column

us_statistics = us_statistics.rename({"Year": "Year of Data Analysis"})

print(us_statistics)
```

```python
#Calculating the mean

us_statistics.select([pl.mean("Net Economic Surplus").alias("mean"),])
```

```python
#Calculating the median

us_statistics.select([pl.mean("Net Economic Surplus").alias("median"),])
```

```python
#Calculating aggregate statistics

us_statistics.describe()
```

```python
#Sorting the data in ascending order by Unadjusted Business Profits

us_statistics.sort("Unadjusted Business Profits")
```

```python
#Sorting the data in descending order by Unadjusted Business Profits

us_statistics.sort("Unadjusted Business Profits", descending=True)

#Note that the correct command is descending=True, not reverse=True as in the Cheat Sheet
```

```python
#Wide to long format (using .inpivot)
#Please note that .melt has been deprecated

us_statistics_long = us_statistics.unpivot(
    on=["Unadjusted Business Profits", "Unadjusted Depreciation Estimates"],
    variable_name="Component",
    value_name="Value",
)

print(us_statistics_long)

#The opposite transformation can be done with .pivot()
```

```python
#Building a pipeline in Polars

(us_statistics
  .filter(pl.col("Year of Data Analysis") > 1945)
  .select(pl.col(["Year of Data Analysis", "Net Economic Surplus"]))
  .sort(pl.col("Net Economic Surplus"), descending=True)
  .slice(0, 5)
)
```
