```python
activities_with_predecessors = {
    'C': {'A'},
    'D': {'A', 'B'},
    'E': {'C'},
    'F': {'C'},
    'G': {'D', 'E'},
    'H': {'F', 'G'}
}

normal_time_duration = {
    'A': 2,
    'B': 3,
    'C': 2,
    'D': 4,
    'E': 4,
    'F': 3,
    'G': 5,
    'H': 2
}

crash_time_duration = {
    'A': 1,
    'B': 1,
    'C': 1,
    'D': 3,
    'E': 2,
    'F': 2,
    'G': 2,
    'H': 1
}

normal_costs = {
    'A': 22000,
    'B': 30000,
    'C': 26000,
    'D': 48000,
    'E': 56000,
    'F': 30000,
    'G': 80000,
    'H': 16000
}

crash_costs = {
    'A': 22750,
    'B': 34000,
    'C': 27000,
    'D': 49000,
    'E': 58000,
    'F': 30500,
    'G': 84500,
    'H': 19000
}

full_activities = set(activities_with_predecessors.keys()) | {'A', 'B'}

processing_order = ('A', 'B', 'C', 'D', 'E', 'F', 'G', 'H')

time_saved = {}
cost_difference = {}
crash_cost_week = {}

for activity in processing_order:
    time_saved[activity] = normal_time_duration[activity] - crash_time_duration[activity]
    cost_difference[activity] = crash_costs[activity] - normal_costs[activity]
    crash_cost_week[activity] = cost_difference[activity] / time_saved[activity]

print(time_saved)
print(cost_difference)
print(crash_cost_week)

#RECALL THAT CRITICAL PATH IS A → C → E → G → H
```
