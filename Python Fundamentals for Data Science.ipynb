Python code is organized into **modules**.

Modules can contain ***functions***, ***classes***, ***variables***, and ***executable statements***.

Functions, classes, and other objects implementing __call__ are callable; calling a class typically creates an instance.

Classes have attributes, including data attributes and methods, which are functions defined in the class.

Packages organize modules and subpackages into namespaces.

Libraries are reusable collections of code, often distributed as one or more modules or packages.

Some of the most common libraries in Python are:

    - NumPy and Pandas (based on NumPy): for arithmetic operations and data analysis

    - Matplotlib and Seaborn (its alternative): for data visualization

Additionally, it is good to keep in mind:

    - os, sys (for internal configurations, such as setting the directory or fixing the time)

    - random (this library is important in Data Science for reproducibility purposes, which is set with random.seed())

Finally, we have some more advanced libraries:

    - Scikit‑learn for Machine Learning

    - Keras, TensorFlow or PyTorch for Deep Learning

Note: we have written the library names in uppercase, but by convention we write them in lowercase in Python.

```python
#LOOPS

#We have 2 main types of Loops: "For" and "While"

#Example of a "For" Loop:

import numpy as np
import matplotlib.pyplot as plt

T = 200
x = np.empty(T + 1)
x[0] = 0
alpha = 0.9

for t in range(T):
    x[t + 1] = alpha * x[t] + np.random.randn()

plt.plot(x)
plt.show()

#In this introductory Python Jupyter Notebook, we are placing the libraries in each "cell"

#By convention in programming, libraries and modules are all grouped together at the beginning - we will do this in future, larger projects
```

```python
#Example of the "While" Loop:

import numpy as np
import matplotlib.pyplot as plt

ts_length = 100

epsilon_values = []

i = 0

while i < ts_length:
    e = np.random.randn()
    epsilon_values.append(e)
    i = i + 1

plt.plot(epsilon_values)

plt.show

#Notice, additionally, the attention to "indentations": it is the way to logically structure what we want to do

#We see that a message appears: "<function matplotlib.pyplot.show(close=None, block=None)>"

#This is because we have used a very simple command ("plt.plot") on its own in which we have not defined all the arguments
```

```python
#STATEMENTS - "If" and "Else"

#Additionally, we have Statements of a more implicit nature (used within arguments), such as "True" or "False"

#They can be used with Loops and other types of objects in Python

#Notice that we are gradually writing longer code. For this reason, it is important to respect the logic of indentations.

import numpy as np
import matplotlib.pyplot as plt

a = 0.9
T = 200
x = np.empty(T+1)
x[0] = 0

for t in range(T):
    if x[t] < 0:
        abs_x = - x[t]
    else:
        abs_x = x[t]
        x[t+1] = a * abs_x + np.random.randn()

plt.plot(x)
plt.show()
```

```python
#Everything we have seen previously can be used "at the same time" for whatever purpose we wish

#Let's look, for example, at a "manual" way to approximate the value of "pi":

n = 100000
count = 0

for i in range(n):
    u, v = np.random.uniform(), np.random.uniform()
    d = np.sqrt((u - 0.5)**2 + (v - 0.5)**2)
    if d < 0.5:
        count += 1

area_estimate = count / n

print(area_estimate * 4) 

#Additionally, note that in this "cell", although we are working with NumPy, we have not loaded the library in this "cell"

#This is because we loaded it in previous "cells".

#For this reason, in large Data Science projects, it makes sense to load it only once in the initial "cell"

#As we said before, the basic arithmetic operators come by "default". (As do Loops and Statements.)

#Thus, we have used NumPy only for the random uniform distribution and the square root
```

```python
#Let's look at a type of class for a very common problem: errors

#We are going to define an error ourselves and analyze why it occurs

class ZeroArgError(Exception):
    pass

def check_zero(old_function):
    """
    Check the argument passed to the function to ensure it is not zero.
    """
    def new_function(arg):
        if arg == 0:
            raise ZeroArgError ("Zero is passed to the argument")
        old_function(arg)
    return new_function

@check_zero
def print_num(num):
    return(num)

print_num(57)
print_num(0)

#Thus, we see that the number 57 does not give any error, but the number 0 does (since we previously defined it to be that way)

#As we have said, errors are very common and the point is only to understand what causes them in order to be able to fix them

#In this specific case, it is enough that the number to be returned is any number other than 0
```

```python
#To make it even clearer what "classes" are, let's look at the following example of currency conversion:

exchange = {"SGD":{"Euro":1.44}}

class Money:  
    
    def __init__(self, amount, currency):
        self.amt = amount
        self.currency = currency     
        
    def __add__(self, money):
        if isinstance(money, self.__class__):
            if self.currency == money.currency:
                return Money(self.amount + money.amt, self.currency)
            else: 
                converted_rate = exchange[self.currency][money.currency]*money.amt
                return Money(self.amt + converted_rate, self.currency)
            
    def __sub__(self, money):
        money.amt*=-1
        return self.__add__(money) 
    
    def convert(self, currency):
        converted_rate = exchange[self.currency][currency]*self.amt
        return Money(converted_rate, currency)
    
    def __repr__(self):
        return "The amount is {} in {}".format(round(self.amt,2),self.currency)
    
Money(20,'SGD') + Money(40,'Euro')

#We see that a single class encapsulates all the functions

#This is the reason (computational efficiency) why in advanced Data Science work, we choose to import only a few specific modules
```
